```python
import requests
from bs4 import BeautifulSoup
```

```python
html_doc = requests.get('https://www.worldometers.info/coronavirus/#google_vignette')
soup = BeautifulSoup(html_doc.content, 'html.parser')
html_doc.status_code
```

```python
table_content = soup.find("table")
```

```python
table_head = table_content.find("thead")
table_body = table_content.find("tbody")
table_row = table_content.find_all('tr')
for data in table_row[8:]:
    print(data.text)
    print('-------------------------------------------------------')
```

```python
headers = []

for th in table_head.find_all('th'):
    headers.append(th.text.strip())

headers
```

```python
rows = []

for tr in table_body.find_all('tr'):
    row = []

    for td in tr.find_all('td'):
        row.append(td.text.strip())

    if row:
        rows.append(row)

rows[8:]
```

```python
import pandas as pd

df = pd.DataFrame(rows, columns=headers)

df.head()
```

```python
for i in range(10):
    print(rows[i][1])  # Country column
```

```python
for row in rows[8:30]:
    print(row[1])
```

```python
unwanted = [
    'World',
    'Europe',
    'Asia',
    'North America',
    'South America',
    'Africa',
    'Oceania'
]

filtered_rows = []

for row in rows:
    if row[1] not in unwanted:
        filtered_rows.append(row)
```

```python
df = pd.DataFrame(filtered_rows, columns=headers)

df.head()
```

```python
df.to_excel("covid_data.xlsx", index=False)
```
