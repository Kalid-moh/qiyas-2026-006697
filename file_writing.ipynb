```python
import string
```

```python

text ="Smoothing is another probabilistic model that combines the document-specific term probability with the global collection probability. The model assigns a weight to each of these components, controlled by a smoothing parameter , which adjusts the balance between the document-specific and global probabilities."
```

```python
table=str.maketrans('','',string.punctuation)
clean_text= text.translate(table)
print(clean_text)
```

```python
sorted_text=sorted(clean_text.lower().split())
sorted_text
```

```python
word_counts = {}

for word in sorted_text:
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

word_counts
```
